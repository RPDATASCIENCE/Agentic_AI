# AI, Machine Learning, Deep Learning, and Generative AI

A practitioner brief on the nested terms that show up in product reviews, architecture docs, and vendor conversations — what each one actually names, where the boundaries are, and how they show up in a live system.

This notebook is **markdown only**: definitions, architecture, and a worked example. It does not call a model.


## Scope

- How **artificial intelligence (AI)**, **machine learning (ML)**, **deep learning (DL)**, and **generative AI (GenAI)** nest — and where they do not
- Discriminative vs generative models; **foundation models** and **large language models (LLMs)**
- The production vocabulary that sits on top: prompt, token, embedding, retrieval-augmented generation (RAG), agent
- When a rule engine, a classifier, or a generator is the right tool
- Common mix-ups that waste design time (ChatGPT ≠ AI, GenAI ≠ all of ML)


## Overview

Teams say “AI” for three different things: a **goal-seeking system**, a **model that learned from data**, and a **product that writes new content**. Those are not synonyms. Treating them as one word leads to the wrong vendor, the wrong risk review, and the wrong success metric.

In production, the useful picture is a **nested capability stack**:

| Layer | What it names | Typical output |
|---|---|---|
| **AI** | A machine-based system that pursues a human-defined objective | A prediction, recommendation, or action |
| **ML** | Techniques that improve that system from data, not only from hand-written rules | A fitted model |
| **Deep learning** | ML that learns layered representations (neural networks) | Features + a decision or generation head |
| **Generative AI** | Models that emit new synthetic content in the shape of the training data | Text, image, audio, code, video |
| **LLM / foundation model** | A large generative (often transformer) model trained on broad data | Next-token predictions you steer with a prompt |

An LLM chat feature is GenAI, which is usually deep learning, which is ML, which is AI. A fraud score from a gradient-boosted tree is ML and AI, but not deep learning and not GenAI. A deterministic “if amount > 10k, hold” rule can still be AI in the statutory sense — and is neither ML nor GenAI.


## Intuition

**The situation.** A stakeholder asks for “some AI” on a product. One person hears a chatbot. Another hears a fraud score. A third hears a flowchart of business rules. The meeting ends with a budget and no shared picture of what will be built.

**The idea.** Think of a new hire on the claims desk. A **script** (“if the receipt is missing, reject”) is intelligence of a kind: it acts toward a goal, but it does not learn. **Training on past claims** is different: the hire starts spotting patterns you never wrote down. If they learn from raw photos and messy notes instead of a spreadsheet of pre-labeled fields, that is a deeper kind of training. If they can **draft a new letter** to the customer, not just stamp approve/deny, they are generating content, not only classifying it.

**Why it matters.** If you asked for a letter-writer and received a stamp, the desk looks idle. If you asked for a stamp and received a letter-writer, you now have fluent text that can invent policy. Cost, latency, and liability follow the layer you actually shipped, not the slide title.

**What we are doing here**

- Name the job: decide, score, or create
- Say whether the system should follow rules, learn from examples, or both
- Say whether the output is a label, a number, or new content
- Keep facts that must be true in a lookup, not in the model’s memory
- Reserve a tool-using loop for work that must act, not only reply

**What we are not doing.** This is not a claim that every product needs a large language model. Most production “AI” is still a classifier or a ranker. Generative systems earn their keep when the user needs new language, images, or code — and they need extra controls because they can sound sure when they are guessing.


## Architecture

![How AI, ML, and generative AI nest](images/ai_ml_genai_terms_architecture.png)

**Nesting (outer to inner)**
1. **Artificial intelligence** — any machine system that pursues a human-defined objective (predict, recommend, or decide). Rules, search, and planning live here even when nothing is learned from data.
2. **Machine learning** — the subset of AI that improves from examples. Tree models and linear models are ML without neural nets.
3. **Deep learning** — the subset of ML that uses multi-layer neural networks. Image and text classifiers live here without generating new content.
4. **Generative AI** — the subset of deep learning that emits new synthetic content (text, image, audio, code).
5. **Large language model** — a common generative-AI instance: next-token models used for chat, drafts, and code.

**Legend:** AI (coral) · Machine learning (magenta) · Deep learning (teal) · Generative AI (inner teal header, white body)

Each inner box is a **subset**, not a synonym. A fraud score can sit in machine learning (or deep learning) and never enter generative AI. A chat feature sits in every ring.


## Concepts

### How the terms nest

```
Artificial intelligence
└── Machine learning          (learns from data; not all AI)
    └── Deep learning         (neural nets with many layers; not all ML)
        └── Generative models (emit new content; not all DL)
            └── Foundation models / LLMs
                └── Applications: prompt, RAG, agents
```

**Not a perfect Russian doll.** Search, planning, and rule engines are AI without ML. Linear models and tree ensembles are ML without deep learning. An ImageNet classifier is deep learning without generation. Some generative methods predate today’s transformers. In 2020s product language, “GenAI” almost always means a **foundation model** that writes text, images, audio, or code.

### Glossary

| Term | Plain meaning | Production tell |
|---|---|---|
| **Artificial intelligence (AI)** | A machine-based system that, for human-defined objectives, makes predictions, recommendations, or decisions | Anything that acts toward a goal, including rules |
| **AI model** | The component that maps inputs to outputs | The artifact you version, evaluate, and roll back |
| **Machine learning (ML)** | Techniques that improve task performance from data | You have a training set and a held-out test set |
| **Supervised learning** | Learn from labeled examples | “This email is spam / not spam” |
| **Unsupervised learning** | Find structure without labels | Clustering customers, compressing features |
| **Self-supervised learning** | Invent a pretext task from the data itself (e.g. predict the next token) | How most LLMs are pre-trained |
| **Reinforcement learning (RL)** | Learn from rewards and penalties, often via trial and error | Game play; some preference-tuning of chat models |
| **Deep learning (DL)** | ML with multi-layer neural networks that learn representations | Pixels, audio, raw text — not only hand-built features |
| **Neural network** | Stacked math units (neurons) with learned weights | The implementation behind most DL |
| **Parameters** | The learned numbers (weights) in a model | Count is a size proxy, not a quality guarantee |
| **Discriminative model** | Predicts a label or score: *P(label \| input)* | Fraud score, intent, ranking |
| **Generative model** | Models how data is formed so it can sample new examples | Draft email, image, code |
| **Generative AI (GenAI)** | AI models that emulate training data in order to generate synthetic content (text, image, audio, video, code) | NIST / CSRC wording; chat and image products |
| **Foundation model** | Trained on broad data, usually self-supervised, reused across tasks | One model, many products via prompts or adapters |
| **Large language model (LLM)** | A (usually transformer) model that predicts the next token of text | Chat, summarization, code, extraction |
| **Transformer** | Neural architecture that uses attention to weigh tokens in a sequence | The dominant LLM backbone since 2017 |
| **Natural language processing (NLP)** | Computational handling of human language | Includes both classifiers and LLMs |
| **Prompt** | The input text (and instructions) that program a generation model | OpenAI: designing a prompt is how you “program” the model |
| **Token** | A chunk of text the model reads or writes | Prompt + output must fit the **context length** |
| **Embedding** | A vector that preserves meaning so similar items sit nearby | Search, clustering, RAG |
| **Retrieval-augmented generation (RAG)** | Fetch relevant documents at query time, then generate with that context | When facts live outside the model weights |
| **Agent** | A model calling tools in a loop until a task is done (model + harness) | LangChain: tools, prompt, and loop around the model |
| **Hallucination** | Fluent output that is not grounded in sources or facts | The default risk of unconstrained generation |
| **Multimodal** | More than one input or output type (text, image, audio) | Screenshot in, text out; image generation |

### How it works (a request)

1. A product frames a user goal as **predict**, **recommend**, or **generate**.
2. If the mapping can be written as stable rules, stop at **AI** without ML.
3. If examples beat rules, **ML** fits a model and you evaluate on held-out data.
4. If the input is high-dimensional (text, image, audio), **deep learning** usually wins.
5. If the user needs **new content**, you call a **generative** model (often an **LLM**).
6. You steer it with a **prompt**; optional **retrieval** supplies facts; an **agent** loop is only for acting with tools.

### When to use which layer

| Job | Prefer | Avoid as the first move |
|---|---|---|
| Stable policy, audit-friendly path | Rules / workflow (still “AI” if it decides) | An LLM that rephrases policy |
| Label, score, rank | Supervised ML (trees or a small net) | A chat model “as the classifier” with no eval |
| Perception (vision, speech, raw text) | Deep learning | Hand-built features unless data is tiny |
| Draft, summarize, extract, code | GenAI / LLM | Training a net from scratch |
| Answers that must match a corpus | RAG (or search + extract) | Hoping the model memorized the PDF |
| Multi-step work with APIs | Agent (tool loop) or an explicit workflow | A single undirected prompt |

### Comparison

| | Rules | Classical ML | Deep learning | Generative AI / LLM |
|---|---|---|---|---|
| Learns from data | No | Yes | Yes | Yes (pre-train + optional adapt) |
| Typical output | Decision | Label / number | Label or embedding | New tokens / pixels |
| Data appetite | None | Hundreds to millions of rows | Large labeled or unlabeled sets | Internet-scale pre-training |
| Failure mode | Blind spots in the spec | Silent mis-scores | Opaque errors | Fluent invention |
| Control | High | Medium (features, thresholds) | Medium (data + eval) | Prompt, tools, retrieval, policy |


## Artificial intelligence

**AI** is the umbrella. U.S. statute (15 U.S.C. 9401(3)) describes a **machine-based system** that, for human-defined objectives, makes **predictions, recommendations, or decisions**, using machine or human inputs, building a model of the environment, and using inference to choose an action.

That definition is wider than “neural nets” and wider than “chatbots.” It includes:

- Rule engines and decision trees written by people
- Optimization and search (routing, scheduling)
- Classical ML and deep learning
- Generative systems

**In a design review, ask:** what objective is the system pursuing, and what action does it take? If you cannot name those, you do not yet have an AI problem statement — you have a slogan.

**AI model** (same family of definitions): the component that implements the mapping from inputs to outputs. Version that artifact. Do not version “the AI” as a vibe.


## Machine learning

**ML** is a set of techniques that **train algorithms from data** so task performance improves without encoding every rule. The product is a **model**; the process is **train → evaluate → deploy → monitor**.

Three setups dominate:

| Setup | You provide | The model learns | Example |
|---|---|---|---|
| **Supervised** | Inputs + labels | A mapping to those labels | Churn yes/no; price; translation pairs |
| **Unsupervised** | Inputs only | Structure (clusters, components) | Segment users; detect odd rows |
| **Reinforcement** | State, actions, reward | A policy that accumulates reward | Bidding; some alignment of chat models |

**Self-supervised** learning sits next to unsupervised: the labels are made from the data (hide a word, predict it). That is how LLM **pre-training** works at scale.

**What ML is not.** A prompt you typed this morning is not “training.” Changing instructions at request time is **inference-time control**. Training updates **parameters** (or adapters) from a dataset and is a release, not a chat turn.


## Deep learning

**Deep learning** is machine learning with **neural networks** that have many layers. Each layer transforms the representation: edges → parts → objects in vision; characters → phrases → intent in language. The network **learns the features** instead of requiring a human to invent them all.

**Why it showed up in products.** Classical ML shines on tables with well-chosen columns. Images, audio, and free text have huge raw dimensionality. Hand-built features plateau; layered networks keep improving with more data and compute.

**Parameters** are the learned weights. Billions of parameters make a model **large**, not automatically **correct**. A 50-parameter linear model with the right features can beat a large net on a small tabular problem.

**Deep learning is still ML:** you need data, a loss, an evaluation set, and a story for drift. The extra cost is data volume, compute, and weaker inherent interpretability.


## Generative AI

NIST’s glossary (CSRC; also NIST AI 600-1 / EO 14110) defines **generative AI** as the class of AI models that **emulate the structure and characteristics of input data in order to generate derived synthetic content** — images, video, audio, text, and other digital content.

**Discriminative vs generative (the useful split):**

| | Discriminative | Generative |
|---|---|---|
| Question | “What class is this?” | “What would a plausible example look like?” |
| Output | Label, probability, rank | New text, image, audio, code |
| Typical use | Fraud, intent, search rank | Drafts, copilots, image tools |

A spam filter and a marketing-copy writer can both be “AI.” Only the second is GenAI.

**Foundation models** are trained on **broad** data, usually with **self-supervision**, and adapted to many tasks. Not all GenAI uses a foundation model (a small GAN that only draws logos is still generative). In enterprise talk, GenAI **usually** means a foundation model plus a product shell (prompt, UI, retrieval, tools).

**Implication for risk.** Discriminative errors are often a wrong number. Generative errors are often a **wrong paragraph that reads right**. Reviews, citations, and retrieval exist because fluency is not a truth signal.


## Large language models and related terms

OpenAI describes **text generation models** (GPT: generative pre-trained transformers) as models trained to understand natural and formal language and to emit text in response to **prompts**. An LLM is a function from text to text: it repeatedly predicts the **next token**.

From that one trick, products hang many jobs: summarize, extract, translate, write code, hold a conversation. The model was not given a separate “summarize” program; it learned patterns that make those behaviors reachable from the right prompt.

| Term | Role next to an LLM |
|---|---|
| **Prompt** | Instructions + user input; the primary control surface at inference |
| **Token** | The unit of context and billing; prompt and completion share a **context length** |
| **Context window** | Maximum tokens of input + output the model can consider at once |
| **Embedding** | Vector for a string; similar meaning → nearby vectors; used for search and RAG |
| **Fine-tuning** | Update weights (or adapters) on your examples; a training release |
| **RAG** | Retrieve passages, then generate with them in the prompt |
| **Agent** | Model + harness: tools in a loop until the task is done |
| **Workflow** | Named, mostly deterministic steps (a graph), with or without an LLM inside a step |

**NLP** is the broader field (language in, structure or language out). LLMs are the current default engine for many NLP products; they did not retire every classifier. Intent routing and PII detection often remain discriminative models sitting **in front of** the LLM.

LangChain’s current split is practical here even if you never import it: an **agent** is a model calling tools in a loop; the **harness** is prompt, tools, and middleware. LangGraph is the **orchestration runtime** when you need named steps, branches, persistence, or a human gate — not when the job is one prompt → one completion.


## How it works end to end

**Offline (learn)**

1. Collect a **corpus** (documents, tickets, images) and, for supervised jobs, **labels**.
2. Choose the layer: rules, tree/linear model, deep classifier, or generative foundation model.
3. Train or adapt. For LLMs this is often **pre-train** (vendor) + optional **fine-tune** (you).
4. Evaluate on held-out data. For GenAI, evaluation is task metrics plus human or LLM-as-judge checks — not only accuracy.

**Online (serve)**

1. User **asks** in the product.
2. The app **frames** the job (classify vs generate vs act).
3. Discriminative path: features or a small net → **score** → policy.
4. Generative path: build a **prompt** (developer instructions + user text + optional retrieved passages).
5. The **LLM** emits tokens until it stops; optional **tools** run in a loop (agent).
6. The product **answers** and logs enough to debug cost, latency, and quality.

OpenAI’s current generation surface for new text apps is the **Responses API** (prompt in, output text or structured data out). Chat Completions still exists; it is not the recommended default for new work. That detail belongs in implementation notebooks; the term-level point is: **generation is next-token prediction steered by a prompt**, regardless of SDK.


## When to use / when not to

**Use the lightest layer that meets the objective.**

- **Rules** when the policy is stable, inspectable, and already known.
- **Classical ML** when you have tabular history and a clear label (churn, fraud, forecast).
- **Deep learning** when the signal is in pixels, waveforms, or raw language and you will invest in data.
- **GenAI / LLM** when the user needs language, code, or other media **created**, not only scored.
- **RAG** when correctness depends on **your** documents, prices, or tickets.
- **Agents** when the model must **do** something (query, ticket, calendar) in a loop, not only write.

**Do not** start with an LLM if a SQL query, a threshold, or a 20-feature classifier already ships the outcome. **Do not** call a classifier “generative AI” in a board deck — you will inherit GenAI risk questions you cannot answer. **Do not** skip evaluation because the demo was fluent.


## Output contract

A term is only useful if the **artifact** is named:

| If you said… | The contract is… |
|---|---|
| AI system | Objective, inputs, action, human override |
| ML model | Training data window, metric, threshold, monitoring |
| Deep model | Input modality, latency/compute budget, eval set |
| GenAI feature | Allowed content types, citation/grounding rule, refusal policy |
| LLM call | Model id, prompt version, token limits, structured vs free text |
| RAG | Corpus, freshness, top-k, what happens when retrieval is empty |
| Agent | Tool list, stop conditions, who may approve irreversible actions |

If two teams cannot fill that row the same way, they are not talking about the same system.


## Walkthrough

Take “summarize this ticket and file a bug if it is a regression.”

1. **AI** — the system has an objective (help engineering) and will recommend or act.
2. **ML / DL** — the summarizer is a neural language model, not a regex.
3. **GenAI** — the summary is **new text**, not a class label.
4. **LLM** — next-token generation from a prompt that includes the ticket.
5. **RAG (optional)** — retrieve similar past bugs so the summary can cite them.
6. **Agent (optional)** — a `create_bug` tool runs only after a confidence or human check.
7. **Not ML training** — editing the prompt is not retraining. Retraining is a later release if examples systematically fail.

If the job were only “is this ticket a regression? yes/no,” a **discriminative** classifier might be enough, with no generation at all.


## Patterns

- **Name the layer in the RFC.** “Supervised ranker” or “LLM with retrieval,” never just “AI.”
- **Separate learn from steer.** Training/fine-tuning vs prompt/RAG vs tool loop are different change-control paths.
- **Ground anything that must be true.** Policy, prices, and account state live in systems of record; the model should read them, not recall them.
- **Keep a discriminative front door.** Language in → intent/PII/jailbreak checks → then generation.
- **Evaluate the job, not the vibe.** Labels and calibration for classifiers; task rubrics and citation checks for GenAI.
- **Prefer an explicit workflow** when the steps are known (validate → retrieve → draft → review). Use an **agent loop** when the tool sequence cannot be written down in advance.


## Pitfalls

| Wrong | Right |
|---|---|
| “We’re adding AI” with no objective | Name predict / recommend / decide / generate |
| ChatGPT used as a synonym for AI | Chat is one GenAI product on an LLM |
| “The model learned that from the prompt” | Prompts steer inference; they do not update weights |
| Calling a fraud XGBoost model GenAI | Discriminative ML; different controls |
| Assuming bigger = better | Match data, latency, and eval; small models win on tables |
| RAG as a magic anti-hallucination switch | Retrieval can be wrong or empty; still measure groundedness |
| Agent when a four-step pipeline would do | LangGraph-style **workflow** vs **agent** is a control-flow choice |
| Shipping Completions-era mental models as current | New OpenAI text apps: **Responses API**; LangChain agents: `create_agent`, not `create_react_agent` |

**Deprecated paths to keep out of new designs:** `LLMChain` / `.run()` / `.predict()`, `langchain-classic` as the default, `from langgraph.prebuilt import create_react_agent`. Those belong in migration notes, not in a glossary of current practice.


## Variants

| You have | Variant |
|---|---|
| Labels and a table | Supervised ML; skip GenAI |
| Lots of unlabeled text, need search | Embeddings + retriever; generation optional |
| Need drafts in your house style | Prompt + few-shot; fine-tune only if prompt fails |
| Need facts from changing docs | RAG; refresh the index on a schedule |
| Need to call internal APIs | Agent (tool loop) or a typed workflow with an LLM in one node |
| Image or audio in or out | Multimodal foundation model, still GenAI |
| On-prem / private weights | Same terms; the **hosting** changed, not the nesting |


## Checklist

- [ ] Objective is written as predict, recommend, decide, or generate
- [ ] Layer chosen: rules / classical ML / DL classifier / GenAI
- [ ] Discriminative vs generative named in the design
- [ ] For ML: train / validation / test split and a production metric
- [ ] For GenAI: prompt owner, model id, token budget, grounding rule
- [ ] For RAG: corpus, update cadence, empty-retrieval behavior
- [ ] For agents: tool allow-list and stop / approval policy
- [ ] Failure mode matches the layer (wrong score vs fluent invention)
- [ ] Monitoring matches the layer (drift and calibration vs groundedness and cost)


## Worked example

**Setting.** A bank operations team wants “AI” on incoming dispute emails.

**Pass 1 — terms, not vendors**

| Need | Layer | Artifact |
|---|---|---|
| Route ATM vs card vs wire | Supervised ML (or rules + ML) | Intent label + confidence |
| Flag likely first-party fraud | Discriminative model | Risk score, threshold, appeal path |
| Draft a customer reply in policy language | GenAI / LLM | Prompt + retrieved policy snippets (RAG) |
| Open a case in the core system | Agent or workflow | Tool call after a human or score gate |

**Pass 2 — what not to build**

- One chat box that “handles disputes” with no routing model: every miss looks like a conversation failure, and you cannot tune routing separately.
- An LLM as the fraud scorer with no calibrated probability: fluent “this looks risky” is not a threshold you can take to model risk.
- Generation **without** retrieval of the current fee schedule: the model will invent fees that used to be true in training data.

**Pass 3 — success metrics by layer**

- Router: precision/recall on intent.
- Fraud: AUC or recall at a fixed false-positive rate; not “the AI sounded careful.”
- Draft: groundedness (policy citations), edit distance vs the sent letter, and time-to-send.
- Tool call: wrong-case rate and approval latency.

The word “AI” still appears once, on the program name. Every workstream is named by **layer and contract**.


## Takeaways

- **AI** is the system that acts toward an objective. **ML** is how some of those systems learn from data. **Deep learning** is ML with stacked neural representations. **Generative AI** emits new content; **LLMs** are the language-shaped instance of that idea.
- Nesting is a default, not a law: rules can be AI; trees can be ML without DL; classifiers can be DL without GenAI.
- **Prompts steer; training updates weights.** Retrieval supplies facts; agents add tools and a loop.
- Pick the **lightest layer** that meets the output contract, then evaluate that layer’s actual failure mode.
- In this repo’s later notebooks, linear generation is an LCEL pipe, a standard tool loop is `create_agent`, and custom control flow is a LangGraph `StateGraph` — those are **application** patterns on top of the terms in this brief.
